### Let's examine some NLP terminologies:
1. corpus: a collection of text and data to train a machine.
2. sentence segmentation: in NLP, to convert the unstructured data into structured, a machine takes a text one sentence at a time. This is called sentence segmentation
3. tokens: small chunks of information that can be classified individually
4. entity: after tokens are created (tokenization), entities are created from each token. An entity is a noun that presents a person, place, thing, or a place.
5. relationship: it is when two or a group of entities that have a strong connection, they can build a relationship.
6. named entity recognition (NER): a technique in NLP that focuses on identifying an classifying entities

```python
# importing general libraries
import numpy as np #np is the alias name for numpy
import pandas as pd # pd is the alias name for pandas
pd.set_option('display.max_rows', 200)


import math
import sys


import spacy
import requests
from bs4 import BeautifulSoup

nlp = spacy.load('en_core_web_sm')
```

```python
# Let's create the data set naming it content

content = "Trinamool Congress leader Mahua Moitra has moved\
the Supreme Court against her expulsion from the Lok Sabha over\
the cash-for-query allegations against her.\
Moitra was ousted from the Parliament last week after the\
Ethics Committee of the Lok Sabha found her guilty of\
jeopardising national security by sharing her parliamentary\
portal's login credentials with businessman Darshan Hiranandani."
```

```python
# let's create an object named doc that contains my data
doc = nlp(content)
```

```python
# let's run a full loop through the date
# to iterate word by word and recognize each entity
for ent in doc.ents:
  print(ent.text, ent.start_char, ent.end_char, ent.label_)
```

**Observation:**
1. The for loop created all the entities with their labels.
2. The for loop iterated from the beginning to the end of the dataset (doc)

```python
# to display the entities with colorful style

# first, we need to import the "displacy" model from spacy library
from spacy import displacy
```

```python
# then create your code for entity display
displacy.render(doc, style= 'ent')
```

```python
# in this step, using a for loop, we create a tabular (rows and columns) output

# first let's create a variable and save the for loop results in it
entities = [(ent.text, ent.label_, ent.lemma_) for ent in doc.ents]
# lemma: is the root of a word
```

```python
# the next step is to create a pandas dataframe (df)
# to have a tabular dataset

df = pd.DataFrame(entities, columns =['Text','Label','Lemma'])
df
```
