- Python Data Structures:
1. List: It is ordered, can have duplicated, mutable (changeable, modifiable)
2. Tuple: It is ordered, can have duplicated, immutable, (cannot change)
3. Set: It is unordered, no duplication, immutable (you can add and remove elements)
4. Dictionary: It is unordered, no duplicated, mutable. A dictionary elements are different from the other data structures element in a way that a dictionary element is a pair of key/value whereas list, tuple, and set have only a single element

```python
# Let's have an example of a list:
L = [1,2,3,4,5,3]
L
```

### 4. Dictionaries
1. There is a big difference between dictionaries and the other 3 data structures,
2. The elements of a dictionary are a pair of key/value.
3. Although a dictionary is unordered (there are no positions), we can index a dictionary and access the elements by the indices (indexes).
4. The data type of a key can't be mutable. It must be only immutable. we can't have a list as a key, but we can have a list as a value)

```python
D = {'a':3.14, 2:'Smith', 'B':19, 5: True}
D
```

```python
type(D)
```

```python
D.keys()
```

```python
D.values()
```

```python
# accessing "Smith" value by its key
D[2]
```

```python
D[5]
```

```python
# changing the value for the key 2 from 'Narges' to 'NLP'
D[True] = "NLP"
D
```

```python
# Let's add a new key and value pair to D
D['newKey'] = 'newValue'
D
```

```python
#adding more than one pair to D
D.update({'c': 'cat', 8:'you', False: 'd'})
D
```

```python
# Let's delete the pair that has 8 as its key
del D[8]
D
```
